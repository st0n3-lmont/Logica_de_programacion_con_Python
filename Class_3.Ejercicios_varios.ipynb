```python
# 17. while con if-else
# Usa un while con if-else para verificar si un numero es par o impar mientras sea menor a 10

num = int(input("Escribe un numero:"))

while num <= 10 :
    if num % 2 == 0 :
        print("El numero es par")
    else :
        print("El numero es impar")
        
    num = int(input("Escribe un numero:"))
# El bucle termina cuando el usuario ingresa un valor mayor a 10
```

```python
# 20. Crear una lista con los primeros 5 numeros pares
# Usa un for para agregar numeros pares a una lista vacia

lista_vacia = []

for k in range(1,11) :
    if len(lista_vacia) <= 5:
        if k % 2 == 0:
            lista_vacia.append(k)
        
print(lista_vacia)
```

```python
# 21 Usa for anidado para imprimir un patrón 
# Usa un for anidado para imprimir un triangulo de asteriscos

for i in range(2,4,2) : # range(limite inf, limite sup, paso)
    for j in range(i) :
        if j % 2 == 0:
            print(" *")
        else :
            print("* *")
```

```python
# 22. Calcular el cuadrado de cada numero en una lista
# Usa un for para imprimir el cuadrado de cada elemento de una lista

lista_num = [1,2,3,4,5,6,7,8,9]
lista_sqrt = []
for i in lista_num :
    j=i**2
    lista_sqrt.append(j)
print(lista_sqrt)
```
