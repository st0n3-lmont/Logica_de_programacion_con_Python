## 9. Eliminar una clave

```python
diccionario_2 = {
    "Electrica": False,
    "Año": 2004,
    "colores:":["azul","rojo","verde"],
    }
print(diccionario_2)
```

```python
del diccionario_2["Electrica"]
print(diccionario_2)
# Se elimina la clave con su valor:
```

## 10. Buscar si una llave existe:

```python
if "Año" in diccionario_2.keys() : 
    print("La clave 'año' esta presente en el diccionario_2")
else : 
    print("La clave 'año' no est apresente en el diccionario_2")
```

## 11. Buscar una llave a partir de un valor

```python
# Asignamos a una variable el valor y otra a la llave
valor_buscado = 2004 

# None es una palabra clave que se utiliza para definir un valor nulo o ningún valor.
# None es diferente de 0, False o una cadena vacía. Acepta ciertas COMPARACIONES, pero no operaciones aritméticas.
llave_buscada = None

# Creamos un ciclo for que va a iterar en cada llave y valor del diccionario 

for clave, valor in diccionario_2.items():
    if valor == valor_buscado:
        llave_buscada = clave
        break
# La siguiente condicion es necesaria para las dos posibilidades: existe o no existe la llave a partir de cierto valor.
        
if clave:
    print(f"Para el valor: {valor_buscado}, la llave es: {llave_buscada}")
else:
    print(f"No se encontro una llave para el valor: {valor_buscado}")
```
