# Diccionarios

## Ejemplo: acceder a los values con las keys

```python
persona = { "nombre" : "jesus" , "edad" : 80 , "ciudad" : "Moscu" }
## Con ayuda de las keys será posible acceder a los valores del diccionario
print(persona["nombre"])
print(persona["edad"])
```

## Dictionary access

## Metodo .get() para acceder al value de una key

```python
print(persona.get("ciudad"))
```

```python
# Si la llvae no existe al correr el codigo no arroja error el metodo get
print(persona.get("Altura"))
```

## Verificar si la clave "Julia" se encuentra en el siguiente diccionario

```python
telefono = { 
            "Jose" : 1234 ,
            "Antonio" : 2345 , 
            "Leonardo" : 3456 }

if "Julia" in telefono:
    print(telefono["Julia"])
else:
    print("Esa clave no existe")
```

## Agrega un elemento extra con su clave - valor al diccionario

```python
telefono["Daniel"] = 6666
print(telefono)
```

## Modifica el numero de "Antonio" a 9876

```python
telefono["Antonio"] = 9876
print(telefono)
```

## Palabra reservada del 
## Elimina un elemento del diccionario (clave-valor)

```python
# La palabra reservada del puede eliminar completamente el diccionario (sin corchetes)
del telefono["Leonardo"]
print(telefono)
```

## Metodo .update()
## Cambia el valor de una llave con .update()

```python
## Consider the following code:
x = {'type' : 'fruit', 'name' : 'banana'}
## Change the "name" "banana" to "apple"
x.update({"name":"apple"})
x
```

## Agrega pares key : value con metodo .update()

```python
x.update({"color":"yellow"})
x
```

## Metodo .pop()
## Elimina un elemento del siguiente dictionary con el metodo .pop()

```python
car =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
car.pop("model")
car
```

## Metodo .clear()
## Use the clear method to empty the cars dictionary.

```python
cars =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
cars.clear()
cars
```
