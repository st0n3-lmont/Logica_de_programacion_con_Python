# Tiempos de espera con diccionarios

## Lista de diccionarios con los tiempos de espera de los pacientes

```python
# Estructura de lista:              lista = [Elemento_0,Elemento_1,]
```

```python
pacientes = [
    {'nombre':'Juan','tiempo_espera':12},
    {'nombre':'Ana','tiempo_espera':25},
    {'nombre':'Luis','tiempo_espera':35},
    {'nombre':'Maria','tiempo_espera':18},
    {'nombre':'Carlos','tiempo_espera': 5}, 
    ]
```

## Extraccion de tiempos de espera:

```python
# Creamos una lista para almacenar los tiempos de espera
t_espera = []
# El ciclo itera en cada elemento de la lista y agrega a la lista vacia cada valor de la key especificada
for i in pacientes:
    t_espera.append(i['tiempo_espera'])
print(t_espera)
```

## Calculo del promedio de los tiempos de espera

```python
tiempo_total = sum(t_espera)
print(tiempo_total)
```

```python
promedio_espera = tiempo_total / len(pacientes)
print(promedio_espera)
```

## Clasificar tiempos de espera 

```python
categorias = {'Corto':0,'Moderado': 0,'Largo':0}
for t in t_espera :
    if t < 15:
        categorias['Corto'] += 1 
    elif 15 <= t <= 30:
        categorias['Moderado'] += 1 
    else:
        categorias['Largo'] += 1
categorias
```

## Mostrar resultados .get() o ciclo for

```python
# print(categorias.get('corto'))
```

```python
# Se usa un ciclo for para imprimir los elementos de un diccionario
# Como cada elemento de un diccionario tiene asigando una clave y un valor  que se iteran con dos variables i,j
# Se utiliza el metodo .items() para llamar a los componentes de cada elemento del diccionario

print(f"Tiempos de espera promedio: {promedio_espera:.2f} minutos")
print("Numero de pacientes en cada categoria:")
print(" ")
for categoria, cantidad in categorias.items():
    print(f"{categoria}: {cantidad} pacientes")
    
```
