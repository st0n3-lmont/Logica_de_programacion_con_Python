## LISTAS POR COMPRESION

```python
# Una lista de 5 numeros por extension
lst = []
for i in range(1,6):
    lst.append(i)
print(lst)
```

```python
# Una lista de 5 numeros por compresion
numeros = [n for n in range(1,6)]
print(numeros)
```

## Crear una lista por comprension que contenga 5 cadenas de caracteres, con uno, dos, tres, cuatro y cinco asteriscos

## Imprime el cuadrado de un numero si dicho numero es par

```python
numeros = [1, 2, 3, 4, 5]
cuadrados = [x ** 2 for x in numeros if x % 2 == 0]
print(cuadrados)
```

## Listas por comprension condicionales

## Extraer los elementos pares de la lista datos 

```python
# Lista por extension
datos = [23,65,78,98,45,4,45,32,65,98]
pares = []
for num in datos:
    if num % 2 == 0:
        pares.append(num)
pares
```

```python
# Lista por comprension
datos = [23,65,78,98,45,4,45,32,65,98]
pares2 = [n for n in datos if n%2 == 0]
pares2
```

## Extraer de la lista datos los numeros mayoes a 30

```python
datos = [23,65,78,98,45,4,45,32,65,98]
# Lista por comprension para filtrar datos
may_30 = [n for n in datos if n > 30]
print(f"datos mayores a 30: {may_30}")
```

## Mas de una condicion

## Extraer los datos mayores a treinta que ademas sean pares

```python
datos = [23,65,78,98,45,4,45,32,65,98]

may_30 = [n for n in datos if n > 30 if n%2==0]

print(f"datos mayores a 30 y ademas son pares: {may_30}")
```

## Crear una lista con las palabras de la lista dada que tengan mas de 7 letras

```python
utensilios = ["mesa","interruptor","silla","microscopio","cubo"]
letras = [palabra for palabra in utensilios if len(palabra) > 7]
letras
```

## Crear una lista con el numero siguiente a los numeros del 1 al 30 que sean multiplos del 7 o multiplos del 11

```python
multiplos_contiguos = [numero + 1 for numero in range(1,31) if numero%7 == 0 or numero%11 == 0]
multiplos_contiguos
```

## Crear una lista con el tipo de datos de cada uno de los elementos que incluye la lista. Si son cadenas: "cadena". Si son int o float: "numerico"

```python
datos_1 = [7, "a",8.2,"s",3]
tipo = ["cadena" if type(dato) == str else
        "numerico" for dato in datos_1]
tipo
```
