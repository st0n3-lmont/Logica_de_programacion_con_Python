## 1. Realiza un programa que pregunte a 5 alumnos si necesitan asesoria y guarda en un diccionario sus nombres y respectiva respuesta.

```python
# Diccionario vacio: {nombres:asesoria}
alum_asesoria = {}

num_alum = int(input("Escribe el numero de alumnos a registrar: "))
j = 0

while j < num_alum :
    alum = str(input("Escribe tu nombre: "))
    asesoria = input(f"{alum} necesitas asesoria? (si/no)")
    alum_asesoria[alum]=asesoria
    
    j += 1
    
print(" ")
print(f"Diccionario con nombres y asesoria: {alum_asesoria}")
```

## 2. Guarda en una lista solo aquellos que respondieron si.

```python
lista_si = []

for nombre,asesoria in alum_asesoria.items() :
    if asesoria.lower() == "si" :
        lista_si.append(nombre)
print(f"Lista de alumnos que necesitan asesoria: {lista_si}")
```

## 3. Genera un diccionario con 5 elementos y guarda en una lista las llaves y y en otra lista los valores de dichos diccionarios.

```python
# Se tiene el diccionario alum_asesoria = {nombres:asesoria} 

nombres = []
asesorias = []

for nombre, asesoria in alum_asesoria.items() :
    nombres.append(nombre)
    asesorias.append(asesoria)

print(f"Lista de nombres (llaves): {nombres} ")
print(f"Lista de asesorias (valores): {asesorias} ")
```
