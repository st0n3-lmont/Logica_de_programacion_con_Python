# Estructuras de datos

Las estructuras de datos nos permiten organizar y almacenar datos de manera eficiente en nuestros programas. Python proporciona varias estructuras de datos integradas, como listas, tuplas, diccionarios y conjuntos, cada una con sus propias características y usos.

# Listas

Una lista es una estructura de datos mutable y ordenada que permite almacenar una colección de elementos. Los elementos de una lista pueden ser de diferentes tipos de datos y se encierran entre corchetes [], separados por comas.

### Creacion y acceso

```python
# Para crear una lista, simplemente encierra los elementos entre corchetes:
frutas = ["manzana", "banana", "naranja"]
```

```python
# Para acceder a los elementos de una lista, utiliza el índice del elemento entre corchetes. Los índices comienzan desde 0.
print(frutas[-1])
```

```python
#También puedes acceder a los elementos desde el final de la lista utilizando índices negativos. 
#El índice -1 representa el último elemento, -2 representa el penúltimo, y así sucesivamente.
print(frutas[-1])
```

### Metodos de listas

Las listas en Python tienen varios métodos incorporados que nos permiten manipular y modificar los elementos de la lista. 
Algunos métodos comunes son:

### .append(elemento)

```python
# Agrega un elemento al final de la lista.
frutas.append("uno")
frutas
```

### .insert(indice,elemento)

```python
# Inserta un elemento en una posición específica de la lista.
frutas.insert(3,"dos")
frutas.insert(3,"dos")
frutas
```

## .remove(elemento)

```python
# Elimina la primera aparición de un elemento en la lista.
frutas.remove("dos")
frutas
```

## .pop(indice)

```python
# Elimina un elemento en una posición específica de la lista.
fruta_eliminada = frutas.pop(2) # elimina "naranja"
print(frutas) 
```

## .sort()

```python
# Ordena los elementos de la lista en orden ascendente.
frutas.sort()
frutas
```

## .reverse()

```python
# Invierte el orden de los elementos en la lista.
frutas.reverse()
frutas
```

## LISTAS POR COMPRENSION

Las listas de comprensión son una forma concisa de crear nuevas listas basadas en una secuencia existente. Permiten filtrar y transformar los elementos de una lista en una sola línea de código.

nueva_lista = [expresion for elemento in secuencia if condicion]

```python
# Imprime el cuadrao de un numero si dicho numero es par
numeros = [1, 2, 3, 4, 5]
cuadrados = [x ** 2 for x in numeros if x % 2 == 0]
print(cuadrados)
```
