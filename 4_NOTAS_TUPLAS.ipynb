# Tuplas

Una tupla es una estructura de datos inmutable y ordenada que permite almacenar una colección de elementos. Los elementos de una tupla se encierran entre paréntesis (), separados por comas.

### Creacion y acceso 

```python
# Estructura de una tupla:                    variable = (posicion_0,posicion_1,...,poscion_n)
punto=(3,4)
# Para acceder a los elementos de una tupla, utiliza el índice del elemento entre corchetes, similar a las listas:
print(punto[0])
```

Las tuplas son útiles cuando necesitas almacenar una colección de elementos que NO deben modificarse, como coordenadas o datos de configuración.

### Metodos de Tuplas

### .count()

```python
tupla = (1,2,3,4,5,6,7,8,9)
tupla.count(2)
# Devuelve la cantidad de veces que tiene contenido un determinado valor
```

### .index() 

```python
tupla_1 = (9,1,2,3,4,5,6,7,8,9)
tupla_1.index(9)
# Devuelve el indice de la PRIMERA aparicion de un elemento en la tupla
```

### .len()

```python
# .len()
len(tupla_1)
```

## Añadir/eliminar elementos*: Convertir la tupla en lista, agregar/modificar elementos.

## Ejercicio clase

```python
# Aplicamos "type cast" a la tupla para convertir a lista

y= list(tupla_1) 


# Agregamos elementos a la lista con .append()
y.append("diez")
print(y)
# Confirmamos que es una lista:
print(type(y))
```

```python
# Usamos el metodo .remove() para eliminar elementos de la lista, NO funciona con la posicion.
y.remove("diez")
y
```

```python
# Intercambiar elementos en funcion del indice sobre la tupla convertida a la lista
y[1]="cero"
y
```

## Buscar en la tupla, lista.

```python
tupla_2 = (1,"dos","3",4,5)
print(tupla_2[1]) ## Solo se especifica el indice y devuelve el valor asignado
```

```python
# Busquedas a trozos (SLICING)
print(tupla_2[1:3]) 
```

## Comprobar si un valor esta dentro de la tupla

```python
print(4 in tupla_2)
```
