## EJERCICIO 1 (#redondeo de flotantes)

```python
salario = float(input("Escribe tu salario $"))

if salario < 18000 :
    
    incremento = salario*(.12)
    print(f"Su incremento del 12% es de: $ {incremento:.2f}")
    salario_inc = salario + incremento
    print(f"Salario con incremento: ${salario_inc:.2f}")
    
elif 18000 <= salario <= 30000:
    
    incremento = salario*(.08)
    print(f"Su incremento del 8% es de: $ {incremento:.2f}")
    salario_inc = salario + incremento
    print(f"Salario con incremento: ${salario_inc:.2f}")
    
elif 30000 < salario <= 50000:
    
    incremento = salario*(.07)
    print(f"Su incremento del 7% es de: $ {incremento:.2f}")
    salario_inc = salario + incremento
    print(f"Salario con incremento: ${salario_inc:.2f}")
    
else:
    
    incremento = salario*(.06)
    print(f"Su incremento del 6% es de: $ {incremento:.2f}")
    salario_inc = salario + incremento
    print(f"Salario con incremento: ${salario_inc:.2f}")

# Nota: Las funciones van a reducir las lineas de código.
```

## EJERCICIO 2

Construye un programa que determine al recibir como datos numeros enteros, si un numero es divisor del otro.

En matemáticas, concretamente en aritmética, se dice que un número entero b es divisible entre otro entero a (no nulo) si al dividirb entre a el resto es cero, o dicho simboloicamente b mod a = 0.

```python
print("Programa que indica que si dados dos numeros uno es divisor del otro")
a=int(input("Escribe el primer numero"))
b=int(input("Escribe el segundo numero"))

if a % b == 0 :
    print(f"{b} es divisor de {a}")
else :
    print(f"{b} no es divisor de {a}")
```

## EJERCICIO 4 (# Forma interesante de hacer incrementos.)

```python
compra = float(input("Ingresa el monto de tu compra: "))
if compra < 800 :
    print("El monto no aplica para el descuento.")
elif 800 <= compra <= 1500 :
    compra -= (compra * 0.10)                                                    # Forma interesante de hacer incrementos.
    print(f"El monto de tu compra con el 10% de descuento: ${compra}")
elif 1500 < compra <= 5000 :
    compra -= (compra * 0.15) 
    print(f"El monto de tu compra con el 15% de descuento: ${compra}")
else:
    compra -= (compra * 0.20) 
    print(f"El monto de tu compra con el 20% de descuento: ${compra}")
```

## EJERCICIO 5

Construye un programa que, al recibir como datos tres numeros reales, identifique cual es mayor. Considera que los numeros pueden ser iguales.

```python
x = int(input("Escribe un numero:"))
y = int(input("Escribe un numero:"))
z = int(input("Escribe un numero:"))

# Casos con repeticion:
if x==y and y==z :
    print("Los numeros son iguales")
elif x==y and y<z :
    print(f"El numero mayor es {z}")  
elif y==z and z<x :
    print(f"El numero mayor es {x}")

# Casos sin repeticion: 
elif y<x and z<x :
    print(f"El numero mayor es: {x} ")
elif x<y and z<y :
    print(f"El numero mayor es: {y} ")
elif x<z and y<z :
    print(f"El numero mayor es: {z} ")
else:
#ultimo caso de repeticion
    print("Caso 331 o 133")
```

## EJERCICIO 6 (Secuencias de de escape (altGr + ? = \ ) con menus usando if)

Construye un programa que permita convertir de pulgadas a milimetros de yardas a metros y de millas a kilometros.
Consideraciones: 
1 pulgada = 25.40 mm
1 yarda = 0.9144 m
1 milla = 1.6093 km

```python
# MENU
opcion = int(input("***MENU*** \n1. Pulgadas a milimetros \n2. Yardas a metros \n3. Millas a kilometros \n\nEscribe una opcion:"))

if opcion == 1 :
    print(f"Opcion {opcion}: Pulgadas a milimetros\n")
    op1 = float(input("Ingresa las pulgadas: "))
    op1 = (op1)*(25.40)
    print(f"{op1:.2f} milimetros")

elif opcion == 2:
    print(f"Opcion {opcion}: Yardas a metros\n")
    op1 = float(input("Ingresa las yardas: "))
    op1 = (op1)*(0.9144)
    print(f"{op1:.2f} metros")

elif opcion == 3:
    print(f"Opcion {opcion}: Millas a kilometros\n")
    op1 = float(input("Ingresa las millas: "))
    op1 = (op1)*(1.6093)
    print(f"{op1:.2f} kilometros")
else:
    print("Ingresa una opcion valida")


    
```

## EJERCICIO 8 (if anidados + seleccion de digitos con posicion)

Construye un programa que, al recibir como dato un numero entero de 4 digitos, determine si todos los DIGITOS son pares.
Por ejemplo, si el numero fuera 5688 no cumpliria la condicion por el digito 5, si por el contrario, el numero fuera 6244 si cumpliria ya que todos los digitos son pares.

```python
num = input("Escribe un numero entero de 4 digitos: ")

if int(num[0]) % 2 == 0:
    if int(num[1]) % 2 == 0:
        if int(num[2]) % 2 == 0:
            if int(num[3]) % 2 == 0:
                print("Todos los digitos son pares")
            else:
                print("El numero contiene digitos impares")
        else:
            print("El numero contiene digitos impares")
    else:
        print("El numero contiene digitos impares")
else:
    print("El numero contiene digitos impares")
            
```
