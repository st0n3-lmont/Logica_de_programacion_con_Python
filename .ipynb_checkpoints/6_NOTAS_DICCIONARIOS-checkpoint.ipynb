# Diccionarios 

# Creacion y acceso

```python
# Para crear un diccionario, utiliza llaves y separa las claves y valores con dos puntos.

persona = {"nombre": "Juan", "edad": 25, "ciudad": "Madrid"}
```

```python
# Para acceder a los valores de un diccionario, utiliza la clave correspondiente entre corchetes:
print(persona["nombre"])
print(persona["ciudad"])
```

# Metodos de diccionarios

# .keys()

```python
# Devuelve una vista de todas las claves del diccionario.
print(persona.keys())
```

# .values()

```python
# Devuelve una vista de todos los valores del diccionario.
print(persona.values())
```

# .items()

```python
# Devuelve una vista de todos los pares clave-valor del diccionario.
print(persona.items())
```

# .update(otro_diccionario)

```python
# Actualiza el diccionario con los pares clave-valor de otro diccionario.
persona.update({"profesion": "Ingeniero"})
print(persona) 
```
