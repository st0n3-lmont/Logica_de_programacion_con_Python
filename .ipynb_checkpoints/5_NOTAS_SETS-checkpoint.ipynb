# Sets (conjuntos) 

Un conjunto es una estructura de datos mutable y NO ORDENADA que permite almacenar una colección de elementos únicos. Los conjuntos se encierran entre llaves {} o se crean utilizando la función set().

### Creacion 

```python
# Para crear conjuntos usa llaves o la funcion set()
frutas = {"manzana", "platano", "pera"}
numeros = set([1,2,3,4,5])

print(frutas)
print(numeros)
```

### Operaciones basicas

Los conjuntos admiten operaciones matemáticas de conjuntos, como la unión (|), la intersección (&), la diferencia (-) y la diferencia simétrica (^).

```python
# Sean los conjuntos:
conjunto1 = {1, 2, 3}
conjunto2 = {3, 4, 5}
```

### Union

```python
union = conjunto1 | conjunto2
print(union)
```

### Interseccion

```python
interseccion = conjunto1 & conjunto2
print(interseccion)
```

### Diferencia

```python
diferencia = conjunto1 - conjunto2
print(diferencia)
```

### Diferencia simétrica

```python
diferencia_simetrica = conjunto1 ^ conjunto2
print(diferencia_simetrica)  
```

## Metodos de conjuntos

## .add()

```python
# Adhiere un elemento al conjunto
conjunto_3 = {"a",1,"cinco",1.2}
conjunto_3.add("tres")
conjunto_3
```

## .remove()

```python
# Remueve un elemento del conjunto
# Si el elemento no existe genera ERROR
conjunto_3.remove('cinco')
conjunto_3
```

## .discard()

```python
# Remueve un elemento del conjunto
# Si el elemento no existe no hace NADA
conjunto_3.discard('siete')
conjunto_3
```

## .clear()

```python
## Elimina todos los elementos de una conjunto
conjunto_3.clear()
conjunto_3
```
