### 1.3 Escribe un programa que permita generar la tabla de multiplicar de un numero entero positivo dado por el usuario. 

```python
num = int(input("Ingresa un numero entero positivo: "))
if num > 0 :
    for j in range(1,11):
        print(num, "Por", j , "es igual a: ", num*j)
else:
    print("Dato erroneo")
```

### 1.4 Repetir un input para dato incorrecto con while 

### Solicitar al usuario escribir un numero positivo, repetir la instruccion hasta que se ingrese el dato solicitado.

```python
# Instrucciones para repetir el ciclo en el caso de un dato incorrecto:
comprob = True
while comprob == True:

    # Instrucciones para pedir el dato al usuario:
    x = int(input("Escribe un numero positivo: "))
    
    # Si se cumple la condicion entonces se detiene el ciclo de repeticion:
    if x > 0:
        comprob = False
        print("El dato ingresado es correcto")
    else:
        print("Dato incorrecto. \n")
```

### 1.5 Escribe un programa que permita generar la tabla de multiplicar de un numero entero positivo dado por el usuario. Si el usuario escribe un numero incorrecto, el programa no se ejecuta. En cambio pregunta de nuevo por la informacion hasta que el dato ingresado sea el correcto.

```python
# Bucle While para repetir la instruccion en caso de que se ingrese un dato invalido

comprobar = True
while comprobar == True:

    #Codigo que genera la tabla de multiplicar
    numb = int(input("Ingresa un numero entero positivo: "))
    if numb > 0 :
        
        comprobar = False                                       # Cuando el usuario ingresa un valor correcto se detiene el while
        
        for j in range(1,11):
            print(numb, "Por", j , "es igual a: ", numb*j)
    else:
        print("El dato ingresado es incorrecto intentalo nuevamente")
```

## EJERCICIO 2

```python
# Instrucciones para repetir el ciclo en el caso de un dato incorrecto:

comprobacion = True
while comprobacion == True:

    # Instrucciones para pedir el dato al usuario:

    nu = int(input("Escribe un numero entero natural: "))
    suma = 0

    # Si se cumple la condicion entonces se detiene el ciclo de repeticion:
    if nu > 0:
        comprobacion = False

# Instrucciones para realizar la suma:

        for s in range(1,nu+1):
            suma += 1/s
        print(f"La suma es: {suma:.3f}")
    else:
        print("Ingresa un dato correcto: ")
```

## EJERCICIO 3

### Construye un programa que al recibir como datos N numeros naturales, determine cuantos de ellos son positivos, negativos o nulos. Si el usuario escribe una cantidad incorrecta a clasificar ,el programa no se ejecuta, en cambio, pregunta de nuevo hasta que el dato ingresado sea el correcto.

```python
# Instrucciones de repeticion si el dato no es un numero natural:

comprob = True
while comprob == True:

# Solicitar la cantidad de numeros a clasificar:
    cantidad = int(input("Escribe cuantos numeros naturales quieres clasificar: "))
    if cantidad > 0 :
        comprob = False   

# Contadores iniciados en 0:
        nulos = 0
        positivos = 0
        negativos = 0

# Instrucciones de conteo
    
        for elemento in range(1,cantidad + 1):
   
    
            numeros = int(input(f"Escribe el numero {elemento}: "))

    # Si no son enteros entonces repetir el ciclo

            if numeros == 0 :
                nulos += 1
            elif numeros > 0 :
                positivos += 1
            elif numeros < 0 :
                negativos += 1
            else:
                print(".")

    else:
            print("Dato incorrecto\n")
        
                
        
print("-----------------------------------------")
print(f"Cantidad de nulos: {nulos}")
print(f"Cantidad de positivoss: {positivos}")
print(f"Cantidad de negativos: {negativos}")
```

## EJERCICIO 4

### Construye un programa que al recibir datos como el peso, altura y el genero de N personas. Obtenga el promedio del peso y el promedio de la altura de ambas poblaciones, masculino y femenino.

```python
personas = int(input("Escribe la cantidad de personas a clasificar:\n"))

# Contadores
mujeres = 0
peso_muj = 0
altura_muj = 0
sum_altura_muj = 0
sum_peso_muj = 0


hombres = 0
peso_hom = 0
altura_hom = 0
sum_altura_hom = 0
sum_peso_hom = 0


x = 0
while x < personas:

    for elemento in range(1,personas + 1):
        genero = input("Escribe tu genero masculino o femenino: ")

        x += 1
    
        if genero.lower() == "masculino" :
            hombres += 1  
            peso_hom = float(input("Escribe tu peso en kg: "))
            sum_peso_hom += peso_hom
            altura_hom = float(input("Escribe tu altura en metros: "))
            sum_altura_hom += altura_hom
            print("\n")
        
        elif genero.lower() == "femenino" :
            mujeres += 1  
            peso_muj = float(input("Escribe tu peso en kg: "))
            sum_peso_muj += peso_muj
            altura_muj = float(input("Escribe tu altura en metros: "))
            sum_altura_muj += altura_muj
            print("\n")
        
        else:
            print("Dato incorrecto")

# Promedios:
# Hombres
prom_peso_hom = sum_peso_hom / hombres
prom_alt_hom = sum_altura_hom / hombres
# Mujeres
prom_peso_muj = sum_peso_muj / mujeres
prom_alt_muj = sum_altura_muj / mujeres


# Resultados con un solo print
print("\nResultados",
      "\nTotal de poblacion: ", personas, "personas",
      "\n",
      "\nTotal mujeres: ", mujeres,
      "\nPromedio altura mujeres: ", prom_alt_muj,
      "\nPomedio peso mujeres: ", prom_peso_muj,
      "\n",
      "\nTotal hombres: ", hombres,
      "\nPromedio altura hombres: ", prom_alt_hom,
      "\nPomedio peso hombres: ", prom_peso_hom )
```

## Ejercicio 5

```python
# Repeticion en caso de tener un dato incorrecto:

rep = True
# while rep = True
# La funcion while primero verifica si el dato es True, entonces simplemente:
while rep:

    number = int(input("Escribe el numero de terminos de la serie:"))

    ans = 0

# El dato es correcto, continua con el proceso y ya no pidas mas datos
    if number > 0 : 
        rep = False 

        # Iteraciones de la serie
        for indice in range(1, number + 1):
            
            # Condicion de numero par
            if indice % 2 == 0:
                ans -= indice**indice
                
            # Condicion de numero impar
            else:
                ans += indice**indice
                
        # Se imprime el resultado afuera del ciclo
        print("El resultado de la serie es: ", ans)
    else :
        print("El dato ingresado es incorrecto, intentalo nuevamente.\n")
```
