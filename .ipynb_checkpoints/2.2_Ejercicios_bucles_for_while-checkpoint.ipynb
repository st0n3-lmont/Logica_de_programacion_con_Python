## Ejercicio 6

### Se dice que un numero N es primo si los únicos enteros positivos que lo dividen son exactamente 1 y N. Construye un programa que lea un numero entero positivo N y escriba todos los numeros primos menores a dicho numero.
### Si el usuario escribe un numero incorrecto, el programa no se ejecuta, en cambio, pregunta de nuevo por la informacion hasta que el dato ingresado sea correcto.

```python
nmbr = int(input("Escribe un numero entero para imprimir todos los numeros primos anteriores:"))

if nmbr > 0 :
    
    
    for indic in range(2,nmbr) : 

        creciente = 2
        primo = True

        while primo and creciente < indic :
            if indic % creciente == 0 :
                primo = False
            else :
                creciente += 1
        if primo == True :
            print(indic, " es primo")



else :
    print(".")
```
