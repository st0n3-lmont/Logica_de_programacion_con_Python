```python
capacidad_hospital=int(input("Ingrese la capacidad del hospital "))
nivel_triage={}
capacidad_triage={}
for i in range(1,6):
  capacidad_triage[i]=int(input(f"Ingrese la capacidad máxima de pacientes del nivel de triage {i} "))
  nivel_triage[i]=int(input(f"Ingrese la cantidad de pacientes hospitalizados del nivel de triage {i} "))
pacientes_hospitalizados=sum(nivel_triage.values())
```

```python
if capacidad_hospital>=pacientes_hospitalizados and nivel_triage[i]<=capacidad_triage[i]:
  while True:
    print("MENÚ PRINCIPAL")
    print("1.Ingresa nuevos pacientes")
    print("2.Ver estado actual del hospital")
    print("3.Salir")
    opcion=int(input("Seleccione una opción "))
    if opcion==1:
      pacientes_nuevos=int (input("Ingrese la cantidad de pacientes nuevos "))
      if pacientes_hospitalizados+pacientes_nuevos> capacidad_hospital:
        print("No hay espacio suficiente en el hospital")
      else:
        for i in range(1,pacientes_nuevos+1):
          nivel=int(input("Ingrese el nivel de triage del paciente "))
          if nivel>=1 and nivel<=5:
            if nivel_triage[nivel]<capacidad_triage[nivel]:
              nivel_triage[nivel]= nivel_triage[nivel]+1
              pacientes_hospitalizados=pacientes_hospitalizados+1
              print(f"Paciente ingresado al nivel de triage {nivel} ")
            else:
              print("No hay espacio suficiente en el nivel de triage seleccionado")
          else:
            print("Nivel de triage inválido")
    elif opcion==2:
      print(f"Estado actual del hospital")
      print(f"Capasidad total del hospital:",{capacidad_hospital})
      print(f"Pacientes hospitalizados:",{pacientes_hospitalizados})
      for i in range(1,6):
        print(f"Nivel de triage {i}: {nivel_triage[i]}")
    elif opcion==3:
      break
    else:
      print("Opción inválida, intentalo nuevamente")
else:
  print("Hay más pacientes de su capacidad en el hospital ó algún nivel esta excedente de su capacidad")
```
