```python
print("Se generó un numero aleatorio entre 0 100\n¡Intenta adivinarlo en 5 intentos!")
# Genera un numero pseaudo-aleatorio
import random 
num_ran = random.randint(1,100)
print("Numero generado para realizar la prueba:", num_ran)

# Ciclo para limitar los intentos del usuario
contador = 1
while contador < 6 :
    num = int(input(f"Intento # {contador} "))
    if (num == num_ran) :
        print("¡Adivinaste!")
        break  
    else :
        print("Intentalo de nuevo")
        contador += 1
        if (contador == 6) :
            print("Tus intentos se han terminado, inicia denuevo el programa.")
            break
```
