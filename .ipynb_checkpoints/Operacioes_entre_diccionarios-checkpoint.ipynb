## 1. Crear un diccionario vacío y uno con valores

```python
dicc_empty = {}
dicc = {
    'name':'antonio',
    'last_name':'Montiel',
    'edad':27,
    'sexo':'masculino'
    }
print(dicc)
```

## Accede y modifica elementos 

## 2. Accede a un valor del diccionario

```python
# Se utiliza la key para acceder al valor
print(dicc['edad'])
```

```python
# O tambien:
print(dicc.get('last_name'))
```

```python
# Extra: type() 
# Type() devuelve el tipo de dato que se esta probando
print(type(dicc['last_name']))
```

## 3. Agrega un nuevo valor de clave - valor al diccionario

```python
dicc['2name'] = 'Alvarez'
print(dicc)
```

## 4. Itera sobre un diccionario (claves-valores)

```python
for i in dicc.items():
    print(i)
```

## 4.1 Itera sobre un diccionario (claves)

```python
for i in dicc.keys():
    print(i)
```

## 4.2 Itera sobre un diccionario (valores)

```python
for i in dicc.values():
    print(i)
```

## 5. Crea una lista de 3 diccionarios que contenga nombre y edad de estudiantes

```python
lista = [
    {'nombre_1':'Antonio','edad_1':27},
    {'nombre_2':'Arturo','edad_2':26},
    {'nombre_3':'Alexis','edad_3':25},
    ]
print(lista)
```

## 6. Imprime el nombre del primer estudiante

```python
# lista = [posicion_1,posicion_2,...,posicion_n]
# En el siguiente ejemplo estaría consultando el primer diccionario, completo: lista[0]
# En seguida consultamos un valor del diccionario contenido en la lista:

print(lista[0]['nombre_1'])
print(lista[2]['edad_3'])
```

## 7. Genera una lista de 2 diccionarios que contenga titulo y autor de libros

```python
biblio = [
    {'Libro':'El Conde MonteCristo','Autor':'Dumas'},
    {'Libro':'El extrangero','Autor':'Camus'}
    ]
print(biblio)
```

```python
# 8. Dado un tercer diccionario con las mismas llaves agregarlo a la lista anterior 
dicc_3 = {'Principia matemathica':'Newton'}
biblio.append(dicc_3)
print(biblio)
```
