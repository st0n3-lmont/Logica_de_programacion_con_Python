# Strings

## Slice Strings

```python
txt = "Hello world"
x = txt[2:5]
print(x)
```

```python
txt = "welcome"
print(txt[3:])
```

## Modify Strings

1.- Convert a string in upper case letter.

```python
x = "welcome"
print(x.upper())
```

2.- Return the string without any white space at the beginig and the end.

```python
txt = " Hello World "
x = txt.strip()
print(x)
```

3.- Convert the value of txt to lower case.

```python
txt_1 = "HELLO WORLD"
txt_1 = txt_1.lower()
print(txt_1)
```

4.- Replace the character H with a J

```python
txt_2 = "Hello World"
txt_2 = txt_2.replace("H","J")
txt_2
```

## Concatenate Strings

1.- What is the correct syntax to merge variable x and y into variable z ?

```python
x = "First"
y = "Second"
z = x + y
print(z)
```

2.- what will be the result of the following syntax ?

```python
x = "Welcome"
y = "Coders"
print(x+y) ## No space on it
```

3.- Consider the following code:

```python
a = 'Join'
b = 'the'
c = 'party'
print(a + ' ' + b + ' ' + c)
```

4.- Consider x = 9, what is a correct syntax to print 'The price is 9.00 dollars' ?

```python
x = 9
print(f"The price is {x:.2f} dollars")
```
