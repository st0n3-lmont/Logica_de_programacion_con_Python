## Verifica si un numero es positivo o negativo usando if-else

```python
x=int(input("Escribe un numero"))
if x < 0 :
    print("El numero es negativo")
else :
    print("El numero es positivo")
```

## Determina si un numero es positivo, negativo o cero usando if-elif-else

```python
y=int(input("Escribe un numero"))
if y > 0 :
    print("El numero es positivo")
elif y < 0 :
    print("El numero es negativo")
else :
    print("El numero es cero")
```

## Usa un for con range para imprimir numeros de 1 al 5

```python
for i in range(1,6) : 
    print(i) 
```

## Imprime cada elemento de una lista usando for

```python
lista_1=[1,2,3,4,5,6,7,8,"nueve"]
for i in lista_1 :
    print(i)
```

## Crea una lista vacia y agrega numeros del 1 al 5 usando .append() en un for

```python
lista_v=[]
for i in range(1,6) :
    lista_v.append(i)
print(lista_v)
```

## Usa un bucle while para imprimir numeros del 1 al 5

```python
j=1
while j <= 5 :
    print(j)
    j += 1
```

# Una clinica quiere hacer un seguimiento de los pacientes que llegan a consulta. Cada dia se registran 
# los nombres de los pacientes que llegan, y se les hace una encuesta para identificar si necesitan una
# encuesta de seguimiento.
# La clinica desea analizar la lista de pacientes para determinar a quienes necesitan recordarles que deben
# Hacer un a cita de seguimiento y cuentos pacientes han asistidoa ella

```python
p_consulta=[]
p_seguimiento=[]

cons = str(input("Si tomaste una cita medica presiona si o no:"))
while cons == "si" : 
    name = str(input("Escribe tu nombre"))
    p_consulta.append(name)
    seg = str(input("Desea agendar una cita de seguimiento? presione si o no:"))
    if seg == "si" : 
        p_seguimiento.append(name)
        print("Has finalizado la encuesta")
        print("-----------------------------------")
        print(" ")
        cons = str(input("Si tomaste una cita medica presiona si o no:"))
    else: 
        print("Has finalizado la encuesta")
        print("-----------------------------------")
        print(" ")
        cons = str(input("Si tomaste una cita medica presiona si o no:"))

print(" ")
print("Resultados: ")
cons_total = len(p_consulta)
print(f"Lista de pacientes que asisitieron a consulta: {p_consulta}")
print(f"Total: {cons_total}")

seg_total = len(p_seguimiento)
print(f"Lista de pacientes que requieren seguimiento: {p_seguimiento}")
print(f"Total {seg_total}")
```
