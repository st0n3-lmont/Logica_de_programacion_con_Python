## Ejercicio: Agenda telefonica con diccionarios

```python
telefonos = {
        "Jose" : 1234,
        "Maria" : 2345,
        "Lucia" : 3456,
        "Andres" : 4567
        }

consultando = True

# Menu
while consultando :
    print()
    print("Mis telefonos")
    print("--------------------")
    print("1.Consultar")
    print("2.Añadir")
    print("3.Modificar")
    print("4.Borrar")
    print("5.Salir")
    print("--------------------")
    
# Pedir al usuario una opcion   
    opcion = ""
    while opcion not in ("1","2","3","4","5"):
        opcion = input ("->")
        
# Consultar
    if opcion == "1":
        nombre = input("Nombre: ")
        if nombre not in telefonos:
            print("Ese nombre no existe")
        else:
            telf = telefonos[nombre]
            print("Numero de ",nombre, ":", telf)
            
# Añadir
    elif opcion == "2":
        nombre = input("Nombre: ")
        if nombre in telefonos:
            print("Ese nombre ya esta en la agenda")
        else:
            telf = input("Telefono: ")
            telefonos[nombre] = telf
            print("El telefono se ha añadido correctamente")

# Modificar
    elif opcion == "3":
        nombre = input("Nombre: ")
        if nombre not in telefonos:
            print("Ese nombre no esta en la agenda")
        else:
            telf = int(input("Telefono: "))
            telefonos[nombre] = telf
            print("El telefono se ha modificado correctamente")

# Borrar
    elif opcion == "4":
        nombre = input("Nombre: ")
        if nombre not in telefonos:
            print("Ese nombre no esta en la agenda")
        else:
            del telefonos[nombre]
            print("El telefono se ha borrado correctamente")

# Salir
    elif opcion == "5":
        consultando = False
```
