```python
capacidad_hospital=int(input("Ingrese la capacidad del hospital "))
pacientes_ingresados=int(input("Ingrese la cantidad de pacientes ingresados"))
nivel_triage={i:0 for i in range(1,6)}
capacidad_traige={1:25,2:25,3:25,4:25,5:25}
while True:
  print("MENÚ PRINCIPAL")
  print("1.Ingresa nuevos pacientes")
  print("2.Ver estado actual del hospital")
  print("3.Salir")
  opcion=int(input("Seleccione una opción"))
  if opcion==1:
    pacientes_nuevos=int (input("Ingrese la cantidad de pacientes nuevos"))
    if pacientes_ingresados+pacientes_nuevos> capacidad_hospital:
      print("No hay espacio suficiente en el hospital")
    else:
      for i in range(1,pacientes_nuevos+1):
        nivel=int(input("Ingrese el nivel de triage del paciente"))
        if nivel>=1 and nivel<=5:
          if nivel_triage[i]< capacidad_traige[i]:
            nivel_triage[i]<- nivel_triage[i]+1
            pacientes_ingresados=pacientes_ingresados+1
            print(f"Paciente ingresado al nivel de triage {nivel}")
          else:
            print("No hay espacio suficiente en el nivel de triage seleccionado")
        else:
          print("Nivel de triage inválido")
  elif opcion==2:
    print(f"Estado actual del hospital")
    print(f"Capasidad total del hospital:",{capacidad_hospital})
    print(f"Pacientes ingresados:",{pacientes_ingresados})
    for i in range(1,6):
      print(f"Nivel de triage {i}: {nivel_triage[i]}")
  elif opcion==3:
    break
  else:
    print("Opción inválida, intentalo nuevamente")
```
