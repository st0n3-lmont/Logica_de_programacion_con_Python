```python
#En una clinica se registran los tiempos de espera (minutos) de varios pacientes
#Cacular el tiempo promedio de espera y clasifica los tiempos en tres categorias:
#Corto: menos de 15 minutos
#Moderado: entre 15 y 30 minutos
#Largo: mas de 30 minutos
#Muestra el promedio de espera y el numero de pacientes en cada categoria.
```

```python
tiempos_espera=[5,10,15,20,35,40,12,25,18,7]
#listas vacias
t_corto=[] # t<15
t_mode=[] # 15<t<30
t_largo=[] # t>30
```

```python
# Promedio general
acum=0
for i in tiempos_espera:
    acum = acum + i
prom=acum/(len(tiempos_espera))
print(f"El tiempo promedio general es: {prom} ")
```

```python
# Tiempos cortos
# Filtrado
for i in tiempos_espera :
    if i <= 15:
        t_corto.append(i)
print(f"los tiempos de espera menores a 15min son: {t_corto}")

# Promedio
suma_c=0
for j in t_corto :
    suma_c = suma_c + j

prom_c = suma_c / len(t_corto)
print(f"Promedio de tiempos menores a 15min: {prom_c} min")
```

```python
# Tiempos moderados
# Filtrado
for i in tiempos_espera :
    if 15 < i <= 30 :
        t_mode.append(i)
print(f"los tiempos de espera entre 15 y 30 min son: {t_mode}")

# Promedio
suma_m=0
for j in t_mode :
    suma_m = suma_m + j

prom_m = suma_m / len(t_mode)
print(f"Promedio de tiempos moderados (15 y 30 min) es: {prom_m} min")
```

```python
# Tiempos largos
# Filtrado
for i in tiempos_espera :
    if i > 30 :
        t_largo.append(i)
print(f"los tiempos de espera entre 15 y 30 min son: {t_largo}")

# Promedio
suma_l=0
for j in t_largo :
    suma_l = suma_l + j

prom_l = suma_l / len(t_largo)
print(f"Promedio de tiempos mayores a 30min: {prom_l} min")
```
