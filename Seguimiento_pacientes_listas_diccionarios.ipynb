## Crear un diccionario con sus claves (pacientes en lista) y solicitar su valor (si requieren seguimiento)

```python
# Lista de pacientes
pacientes_dia = ['alfredo','juan','pedro']
# Diccionario vacio 
pacientes_seguimiento = {}
# Inicializaicion del contador 
seguimiento = 0
```

```python
# El ciclo for itera sobre la lista 
for paciente in pacientes_dia:
    
    # El input solicita en cada iteracion si cada paciente de la lista necesita seguimiento 
    # El input se guarda en la variable que se va a comparar
    necesita_seguimiento = input(f'¿{paciente} necesita consulta de seguimiento? (si/no): ')

    # Se reduce a minusculas la cadena contenida en la variable y se compara el input
    if necesita_seguimiento.lower() == "si" :

        # En seguida se toma el indice de la iteracion (llave) y se guarda con su respectiva respuesta (valor) en el diccioario
        pacientes_seguimiento[paciente] = 'si'
        seguimiento = seguimiento + 1
    else :
        pacientes_seguimiento[paciente] = 'no'

print(" ")
print(f'Totales: {pacientes_seguimiento}')
print(f'Total pacientes que necesitan seguimiento: {seguimiento}')
```

## Crear un diccionario que contenga nombre de paciente (llave) y si necesita cita de seguimiento (valor) solicitando los datos

```python
pacientes_dia_2 = []
pacientes_seguimiento_2 = []
pacientes_no_seguimiento = []

while True:
    paciente  = input('Ingrese el nombre del paciente o "fin" para terminar: ')
    if paciente.lower() == "fin" :
        break
    pacientes_dia_2.append(paciente)
    
# Identificar pacientes que necesitan seguimiento
for paciente in pacientes_dia_2 :
    necesita_seguimiento_2 = input(f"¿{paciente} necesita consulta de seguimiento? (si/no): ")
    if necesita_seguimiento_2.lower() == "si" :
        pacientes_seguimiento_2.append(paciente)
    elif necesita_seguimiento_2.lower() == "no" :
        pacientes_no_seguimiento.append(paciente)

print(f"lista de pacientes que necesitan seguimiento: {pacientes_seguimiento_2}")
print(f"lista de pacientes que NO necesitan seguimiento: {pacientes_no_seguimiento}") 
```
